```python
import sys 
from pathlib import Path
import matplotlib.pyplot as plt
import cv2

# SCR visible
project_root = Path().resolve().parents[1]
sys.path.append(str(project_root))
from src.data.preprocess_functions import preprocess_image

# Normal lungs
path = "../raw/normal/Normal-1.png"
# Loading data
normal = cv2.imread(path,cv2.IMREAD_GRAYSCALE)
preposs = preprocess_image(path)


# Side-by side comparation
fig, ax = plt.subplots(1,2, figsize=(10,4))

ax[0].imshow(normal, cmap="gray")
ax[0].set_title("Original")
ax[0].axis("off")

ax[1].imshow(preposs, cmap="gray")
ax[1].set_title("Preprocessed")
ax[1].axis("off")

plt.savefig(f"results/preprocessing/NormalOriginal_vs_Prepocessed.png",dpi = 300, bbox_inches = "tight")
plt.show()

# Histogram comparation
plt.figure(figsize=(8,4))
plt.hist(normal.flatten(), bins = 50, alpha = 0.6, label = "Original")
plt.hist(preposs.flatten(), bins = 50, alpha = 0.6, label = "Preprocessed")
plt.legend()
plt.title("Pixel Intensity Distribution")
plt.savefig(f"results/preprocessing/Pixel_distributions.png",dpi = 300, bbox_inches="tight")
plt.show()

# ------ Tubercolosis analysis ---------------

path2 = "../raw/tuberculosis/Tuberculosis-10.png"
# Loading data
tuberculosis = cv2.imread(path2,cv2.IMREAD_GRAYSCALE)
tuber_preposses = preprocess_image(path2)


# Side- by side comparation
fig, ax = plt.subplots(1,2, figsize=(10,4))

ax[0].imshow(tuberculosis, cmap ="gray")
ax[0].set_title("Original")
ax[0].axis("off")

ax[1].imshow(tuber_preposses, cmap ="gray")
ax[1].set_title("Preprocessed")
ax[1].axis("off")

plt.savefig(f"results/preprocessing/Tuberculosis_vs_Prepocessed.png",dpi = 300, bbox_inches = "tight")
plt.show()

# Histogram comparation
plt.figure(figsize = (8,4))
plt.hist(tuberculosis.flatten(), bins = 50, alpha = 0.6, label = "Tuberculosis")
plt.hist(tuber_preposses.flatten(), bins = 50, alpha = 0.6, label = "Preprocessed")
plt.legend()
plt.title("Pixel Intensity Distribution")
plt.savefig(f"results/preprocessing/Tuberculosis_pixel_distribution.png",dpi = 300, bbox_inches = "tight")
plt.show()
```
